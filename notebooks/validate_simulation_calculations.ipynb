```python
import plotly.graph_objects as go
from ball_flight import BallPathSimulator
```

```python
sim = BallPathSimulator("Ball Path with Spin")
```

```python
t_max = 5
t_step = 0.005
args = dict(
    v0 = 22, 
    theta = 0, 
    phi = 2, 
    spin_rate = 1800, 
    spin_elevation = 0,  
    spin_azimuth = 0, 
    x0 = -0.5, 
    y0 = 1.6, 
    z0 = 0
    )
args0 = {**args, "spin_rate": 0}
sim.initialize_simulation(**args)
sim.calculate_path(t_max, t_step)
```

```python
fit = go.Scatter(x=sim.x,y=sim.y, mode="lines")
layout = go.Layout(
    title="Fit",
    xaxis_title="X (pixels)",
    yaxis_title="Y (pixels)",
    )

fig = go.Figure(data=[fit], layout=layout)
fig.show(renderer="notebook")
lift = sim.lift_force
```
